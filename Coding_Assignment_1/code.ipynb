```python
# Disease Prediction using Logistic Regression
# and SVM Classification

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    classification_report
)

print("Libraries imported successfully!")

# Load Breast Cancer Wisconsin Diagnostic Dataset
data = load_breast_cancer()

# Create DataFrame
X = pd.DataFrame(data.data, columns=data.feature_names)

# Target variable
y = pd.Series(data.target, name="Disease")

print("Dataset loaded successfully!")
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])

print("\nTarget classes:")
print(data.target_names)

print("\nFirst 5 rows:")
display(X.head())

# Dataset information
print("Dataset Shape:", X.shape)

print("\nClass Distribution:")
print(y.value_counts())

# Display class names
print("\nTarget Encoding:")
for i, name in enumerate(data.target_names):
    print(i, "=", name)

# Convert target:
# Malignant = 1 (positive class)
# Benign = 0 (negative class)

y_medical = (y == 0).astype(int)

print("New target distribution:")
print(y_medical.value_counts())

print("\n0 = Benign")
print("1 = Malignant")

# Check missing values
missing_values = X.isnull().sum().sum()

print("Total missing values:", missing_values)

if missing_values == 0:
    print("There are no missing values.")
else:
    print("Missing values detected.")
# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_medical,
    test_size=0.20,
    random_state=42,
    stratify=y_medical
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

# Standardize features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed.")
# Create Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=10000,
    random_state=42
)

# Train the model
logistic_model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
# Predict test data
y_pred_lr = logistic_model.predict(X_test_scaled)

print("Predictions generated.")
print("First 20 predictions:")
print(y_pred_lr[:20])

# Confusion Matrix
cm_lr = confusion_matrix(y_test, y_pred_lr)

print("Logistic Regression Confusion Matrix:")
print(cm_lr)

# Plot confusion matrix
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Benign", "Malignant"],
    yticklabels=["Benign", "Malignant"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression - Confusion Matrix")
plt.show()
# Extract TN, FP, FN, TP
TN, FP, FN, TP = cm_lr.ravel()

# Calculate metrics
accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr = precision_score(y_test, y_pred_lr)
recall_lr = recall_score(y_test, y_pred_lr)

# Specificity = TN / (TN + FP)
specificity_lr = TN / (TN + FP)

print("==========================================")
print("LOGISTIC REGRESSION PERFORMANCE")
print("==========================================")

print(f"Accuracy    : {accuracy_lr:.4f}")
print(f"Precision   : {precision_lr:.4f}")
print(f"Recall      : {recall_lr:.4f}")
print(f"Sensitivity : {recall_lr:.4f}")
print(f"Specificity : {specificity_lr:.4f}")
print("Classification Report - Logistic Regression")
print(
    classification_report(
        y_test,
        y_pred_lr,
        target_names=["Benign", "Malignant"]
    )
)

# Create SVM classifier
svm_model = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    random_state=42
)

# Train SVM
svm_model.fit(X_train_scaled, y_train)

print("SVM model trained successfully!")

# Predict test data
y_pred_svm = svm_model.predict(X_test_scaled)

print("SVM predictions generated.")
print("First 20 predictions:")
print(y_pred_svm[:20])

# Confusion Matrix
cm_svm = confusion_matrix(y_test, y_pred_svm)

print("SVM Confusion Matrix:")
print(cm_svm)

# Plot confusion matrix
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_svm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Benign", "Malignant"],
    yticklabels=["Benign", "Malignant"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("SVM - Confusion Matrix")
plt.show()

# Extract TN, FP, FN, TP
TN, FP, FN, TP = cm_svm.ravel()

# Calculate metrics
accuracy_svm = accuracy_score(y_test, y_pred_svm)
precision_svm = precision_score(y_test, y_pred_svm)
recall_svm = recall_score(y_test, y_pred_svm)

# Specificity
specificity_svm = TN / (TN + FP)

print("==========================================")
print("SVM PERFORMANCE")
print("==========================================")

print(f"Accuracy    : {accuracy_svm:.4f}")
print(f"Precision   : {precision_svm:.4f}")
print(f"Recall      : {recall_svm:.4f}")
print(f"Sensitivity : {recall_svm:.4f}")
print(f"Specificity : {specificity_svm:.4f}")

print("Classification Report - SVM")

print(
    classification_report(
        y_test,
        y_pred_svm,
        target_names=["Benign", "Malignant"]
    )
)

# Create comparison DataFrame

results = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall / Sensitivity",
        "Specificity"
    ],

    "Logistic Regression": [
        accuracy_lr,
        precision_lr,
        recall_lr,
        specificity_lr
    ],

    "SVM": [
        accuracy_svm,
        precision_svm,
        recall_svm,
        specificity_svm
    ]
})

# Display results
display(results)

results_percentage = results.copy()

results_percentage["Logistic Regression"] = (
    results_percentage["Logistic Regression"] * 100
).round(2)

results_percentage["SVM"] = (
    results_percentage["SVM"] * 100
).round(2)

print("Model Performance Comparison (%)")
display(results_percentage)
# Prepare data for plotting
plot_data = results.set_index("Metric")

# Plot
ax = plot_data.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Logistic Regression vs SVM")
plt.xlabel("Evaluation Metric")
plt.ylabel("Score")
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(title="Model")
plt.grid(axis="y", alpha=0.3)

plt.show()

print("=" * 60)
print("FINAL MODEL COMPARISON")
print("=" * 60)

print("\nLogistic Regression:")
print(f"Accuracy              : {accuracy_lr * 100:.2f}%")
print(f"Precision             : {precision_lr * 100:.2f}%")
print(f"Recall / Sensitivity  : {recall_lr * 100:.2f}%")
print(f"Specificity           : {specificity_lr * 100:.2f}%")

print("\nSVM:")
print(f"Accuracy              : {accuracy_svm * 100:.2f}%")
print(f"Precision             : {precision_svm * 100:.2f}%")
print(f"Recall / Sensitivity  : {recall_svm * 100:.2f}%")
print(f"Specificity           : {specificity_svm * 100:.2f}%")

print("=" * 60)
```
